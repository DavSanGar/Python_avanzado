2. Crea una clase llamada "Calculadora" con métodos para sumar, restar,
multiplicar y dividir. Crea objetos de esta clase y realiza algunas operaciones
básicas.

```python
class calculadora:
    def sumar(self, a, b):
        return a + b
    
    def restar(self, a ,b):
        return a - b
    
    def multiplicar(self, a, b):
        return a * b
    
    def dividir(self, a, b):
        return a / b

calculadora = calculadora() 
print("El resultado de la suma es: ", calculadora.sumar(5,3))
print("El resultado de la resta es: ", calculadora.restar(5,3))
print("El resultado de la multiplicación es: ", calculadora.multiplicar(5,3))
print("El resultado de la división es: ", calculadora.dividir(5,3))
```

4. Crea una clase "Rectangulo" con atributos de longitud y ancho. Implementa
un método para calcular el área y el perímetro del rectángulo.

```python
class rectangulo:
    def __init__(self, longitud, ancho):
        self.longitud = longitud
        self.ancho = ancho
    
    def area_rectangulo(self):
        area = self.longitud * self.ancho
        return area
    
    def perimetro(self):
       return 2 * (self.longitud + self.ancho)
    
rectangulo1 = rectangulo(5,2)
print("Area", rectangulo1.area_rectangulo())
print("Perímetro", rectangulo1.perimetro())


```

4. Crea una clase "Dado" que simule el lanzamiento de un dado de 6 caras.
Implementa un método para lanzar el dado y mostrar el resultado (quizás te
convenga usar el modulo random).

```python
import random

class dado:
    def lanzar(self):
        lanzamiento = random.randint(1,6)
        return lanzamiento
    
dado1 = dado()
dado2 = dado()
resultado1 = dado1.lanzar()
resultado2 = dado2.lanzar()
print("El resultado del dado1 es ", resultado1, "El resultado del dado2 es ", resultado2)
```

5. Crea una clase "Coche" con atributos como marca, modelo y año.
Implementa un método para encender el coche y otro para apagarlo (puedes
simulae el encendido y apagado con una variable booleana).

```python
class coche:
    def __init__(self, marca, modelo, año):
        self.marca = marca
        self.modelo = modelo
        self.año = año
        self.encendido = False
    
    def encender(self):
        self.encendido = True

    def apagar(self):
        self.encendido = False

mi_coche = coche("Hyundai", "i30", 2012)
print("Mi coche es un",mi_coche.marca, mi_coche.modelo, "del año", mi_coche.año)
print("El coche esta encendido ", mi_coche.encendido)
mi_coche.encender()
print("El coche está encendido: ", mi_coche.encendido)
mi_coche.apagar()
print("El coche está encendido: ", mi_coche.encendido) 
```
