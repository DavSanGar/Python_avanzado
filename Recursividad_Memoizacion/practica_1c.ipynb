FACTORIAL
Implementa una función recursiva llamada factorial que calcule el factorial de
un número entero positivo. El factorial de un número n se define como el
producto de todos los números enteros positivos desde 1 hasta n.
(El factorial de 0 y de 1 es igual a 1)

```python
def factorial(n):
    '''Calcula el factorial de n
    INPUT:
    - n: int
    OUTPUT:
    - resultado: int'''
    
    #Caso base
    if n == 0 or n == 1:
        return 1
    
    #plantear sentencia recursiva
    else:
        return n * factorial(n-1)

#caso de uso
factorial(7)
print(factorial(7))
'''
7 * factorial(6)
7 * 6 * factorial (5)
7 * 6 * 5 * factorial(4)
7 * 6 * 5 * 4 factorial(3)
7 * 6 * 5 * 4 * 3 * factorial(2)
7 * 6 * 5 * 4 * 3 * 2 * factorial(1)
7 * 6 * 5 * 4 * 3 * 2 * 1 
'''
```

POTENCIA
Implementa una función recursiva llamada potencia que calcule el resultado
de elevar un número a una potencia dada. Puedes asumir que tanto el
número como la potencia son enteros no negativos.

```python
def potencia(base, exponente):
    '''Calcula la potencia de un número'''
    #INPUT:
    # - base: el numero que queremos elevar (int positivo)
    # - exponente: int positivo
    #OUPUT: 
    # - rersultado: int positivo

    #caso base
    if exponente ==  0:
        return 1
    
    #sentecia recursiva
    else:
        return base * potencia(base, exponente - 1)
    
potencia(5,5)

''' 
5 * potencia(5,4)
5 * 5 * potencia(5,3)
5 * 5 * 5 * potencia(5,2)
5 * 5 * 5 * 5 * potencia(5,1)
5 * 5 * 5 * 5 * 5 * potencia(5,0)
5 * 5 * 5 * 5 * 5 * 1

'''
```

SUMA ELEMENTOS DE UNA LISTA
Crea una función recursiva llamada suma_lista que calcule la suma de todos
los elementos de una lista de enteros. Puedes asumir que la lista no está
vacía.

```python
def suma_lista(lista):
    '''Suma los elementos de la lista'''
    #INPUT:
    # - lista: lista numeros enteros
    # OUTPUT:
    #resultado: suma de los elementos de la lista

    #caso base:
    if len(lista) == 1:
        return lista[0]
    
    #sentecia recursiva:
    else:
        return lista[0] + suma_lista(lista[1:])
    
    #caso de uso:
numeros = [1,2,3,4,5]
suma_lista(numeros)

'''
1 + suma_lista([2,3,4,5])
1 + 2 + suma_lista([3,4,5])
1 + 2 + 3 + suma_lista([4,5])
1 + 2 + 3 + 4 suma_lista([5])
1 + 2 + 3 + 4 + 5 
'''
```

NUMERO TRIANGULAR
Crea una función recursiva llamada numero_triangular que calcule el n-ésimo
número triangular. Un número triangular se obtiene al sumar todos los
números naturales desde 1 hasta n.

```python
def numero_triangular(n):
    '''Calcula el valor del numero triangular de n'''
    #INPUT:
    # - n: int positivo
    # OUTPUT:
    #resultado: suma de valores para obtener numero triangular

    #caso base:
    if n == 1:
        return 1
    
    #sentecia recursiva:
    else:
        return n + numero_triangular(n-1)
    
#caso de uso:
numero_triangular(4)

'''
4 + numero_triangular(3)
4 + 3 + numero_triangular(2)
4 + 3 + 2 numero_triangular(1)
4 + 3 + 2 + 1
'''
```
